```python
import pandas as pd
df = pd.read_csv('games2cleaned.csv')

# Dimensiones del dataset
print(df.shape)

# Tipos de datos y nulos
print(df.info())

# Cuántos nulos por columna
print(df.isnull().sum())
```

```python
# Ver ejemplos reales de developers, publishers, genres, categories
print(df['developers'].value_counts().head(10))
print(df['publishers'].value_counts().head(10))
print(df['genres'].value_counts().head(10))
print(df['categories'].value_counts().head(10))

# ¿Cuántos juegos tienen MÁS DE UN developer?
df['num_devs'] = df['developers'].apply(lambda x: len(eval(x)) if pd.notna(x) else 0)
print(df['num_devs'].value_counts())
```

```python
# Ver formato exacto
print(df['estimated_owners'].value_counts())
```

```python
# Celda 4 corregida
df['release_date'] = pd.to_datetime(df['release_date'])
df['release_year'] = df['release_date'].dt.year

print(df['release_year'].value_counts().sort_index())
```

```python
# Celda 5 - Entender mejor genres y categories por dentro
import ast

# Cuántos géneros distintos hay en total
all_genres = []
for x in df['genres'].dropna():
    try:
        all_genres.extend(ast.literal_eval(x))
    except:
        pass
print("Géneros únicos:", pd.Series(all_genres).nunique())
print(pd.Series(all_genres).value_counts().head(15))

# Cuántas categorías distintas hay
all_cats = []
for x in df['categories'].dropna():
    try:
        all_cats.extend(ast.literal_eval(x))
    except:
        pass
print("\nCategorías únicas:", pd.Series(all_cats).nunique())
print(pd.Series(all_cats).value_counts().head(15))
```

```python
# Celda 6 - Métricas básicas de las medidas
cols = ['price', 'positive', 'negative', 'recommendations', 
        'average_playtime_forever', 'peak_ccu', 'pct_pos_total',
        'metacritic_score', 'achievements', 'dlc_count']
print(df[cols].describe())
```

```python
# Celda 7 - Juegos gratuitos vs de pago
print("Gratuitos:", df[df['price'] == 0].shape[0])
print("De pago:", df[df['price'] > 0].shape[0])
print("\nPrecio medio juegos de pago:", df[df['price'] > 0]['price'].mean().round(2))
print("Precio máximo:", df['price'].max())
```

```python
# Celda 8 - Ver publishers y developers vacíos más en detalle
print("Developers vacíos []:", df[df['developers'] == '[]'].shape[0])
print("Publishers vacíos []:", df[df['publishers'] == '[]'].shape[0])

# Cuántos developers únicos hay
import ast
all_devs = []
for x in df['developers']:
    try:
        all_devs.extend(ast.literal_eval(x))
    except:
        pass
print("\nDevelopers únicos:", pd.Series(all_devs).nunique())
print("Total registros developer:", len(all_devs))

all_pubs = []
for x in df['publishers']:
    try:
        all_pubs.extend(ast.literal_eval(x))
    except:
        pass
print("\nPublishers únicos:", pd.Series(all_pubs).nunique())
print("Total registros publisher:", len(all_pubs))
```

```python
import pandas as pd

df = pd.read_csv('games2cleaned.csv')

# Calcular release_year primero
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
df['release_year'] = df['release_date'].dt.year

# Muestra estratificada por año
df_sample = df.groupby('release_year', group_keys=False).apply(
    lambda x: x.sample(min(len(x), int(10000 * len(x) / len(df))))
)

# Exactamente 10000
df_sample = df_sample.sample(min(len(df_sample), 10000), random_state=42)

# Quitar release_year que añadimos nosotros (no estaba en el original)
df_sample = df_sample.drop(columns=['release_year'])

print(df_sample.shape)
print(df_sample['release_date'].dt.year.value_counts().sort_index())

# Guardar
df_sample.to_csv('games2cleaned_10k.csv', index=False)
print("CSV guardado correctamente")
```

```python
import pandas as pd
import random
import ast

random.seed(42)

df = pd.read_csv('games2cleaned_10k_new.csv')

# Extraer developers únicos
all_devs = set()
for x in df['developers']:
    try:
        devs = ast.literal_eval(str(x))
        all_devs.update(devs)
    except:
        pass

# Quitar vacíos
all_devs.discard('')

# Contar juegos por developer para asignar size
dev_game_count = {}
for x in df['developers']:
    try:
        devs = ast.literal_eval(str(x))
        for d in devs:
            d = d.strip()
            if d:
                dev_game_count[d] = dev_game_count.get(d, 0) + 1
    except:
        pass

countries = ['USA', 'UK', 'Japan', 'Germany', 'France',
             'Canada', 'Russia', 'Poland', 'Sweden', 'Australia']
weights   = [0.30, 0.12, 0.10, 0.08, 0.07,
             0.07, 0.07, 0.06, 0.07, 0.06]

rows = []
dev_id = 1
for dev in sorted(all_devs):
    count = dev_game_count.get(dev, 1)
    if count == 1:
        size = 'indie'
    elif count <= 5:
        size = 'mid'
    else:
        size = 'aaa'

    rows.append({
        'developer_id':   dev_id,
        'name':           dev,
        'country':        random.choices(countries, weights=weights, k=1)[0],
        'founding_year':  random.randint(1975, 2022),
        'size':           size,
        'is_closed':      random.random() < 0.05
    })
    dev_id += 1

df_devs = pd.DataFrame(rows)
print(df_devs.shape)
print(df_devs['size'].value_counts())
print(df_devs['country'].value_counts())
print(df_devs.head(5))

df_devs.to_csv('developers.csv', index=False)
print("developers.csv guardado correctamente")
```

```python
# Extraer publishers únicos
all_pubs = set()
for x in df['publishers']:
    try:
        pubs = ast.literal_eval(str(x))
        all_pubs.update(pubs)
    except:
        pass

all_pubs.discard('')

pub_game_count = {}
for x in df['publishers']:
    try:
        pubs = ast.literal_eval(str(x))
        for p in pubs:
            p = p.strip()
            if p:
                pub_game_count[p] = pub_game_count.get(p, 0) + 1
    except:
        pass

rows = []
pub_id = 1
for pub in sorted(all_pubs):
    count = pub_game_count.get(pub, 1)
    if count == 1:
        size = 'indie'
    elif count <= 5:
        size = 'mid'
    else:
        size = 'aaa'

    rows.append({
        'publisher_id':  pub_id,
        'name':          pub,
        'country':       random.choices(countries, weights=weights, k=1)[0],
        'founding_year': random.randint(1975, 2022),
        'size':          size,
        'is_closed':     random.random() < 0.05
    })
    pub_id += 1

df_pubs = pd.DataFrame(rows)
print(df_pubs.shape)
print(df_pubs['size'].value_counts())
print(df_pubs.head(5))

df_pubs.to_csv('publishers.csv', index=False)
print("publishers.csv guardado correctamente")
```

```python
import pandas as pd
import random

random.seed(42)

df = pd.read_csv('games2cleaned_10k_new.csv')

# Lógica de asignación de plataformas:
# - Todos los juegos tienen Steam (1)
# - Top juegos por peak_ccu también en Epic (2) y GOG (3)
# - Juegos indie pequeños pueden estar en itch.io (4)
# - Solo los más grandes en Amazon Luna (5)

rows = []

for _, row in df.iterrows():
    app_id = int(row['appid'])
    peak_ccu = int(row['peak_ccu']) if str(row['peak_ccu']).isdigit() else 0
    price = float(row['price'])
    
    # Todos tienen Steam
    platforms = [1]
    
    # Top juegos (peak_ccu > 10000) también en Epic y GOG
    if peak_ccu > 10000:
        platforms.append(2)  # Epic
        platforms.append(3)  # GOG
    
    # Juegos pequeños y gratuitos o baratos pueden estar en itch.io
    elif price <= 1.99 and random.random() < 0.3:
        platforms.append(4)  # itch.io
    
    # Solo los más grandes en Amazon Luna (peak_ccu > 100000)
    if peak_ccu > 100000:
        platforms.append(5)  # Amazon Luna

    for platform_id in platforms:
        rows.append({
            'app_id':      app_id,
            'platform_id': platform_id
        })

df_gp = pd.DataFrame(rows)
print(df_gp.shape)
print(df_gp['platform_id'].value_counts())
print(df_gp.head(10))

df_gp.to_csv('game_platforms.csv', index=False)
print("game_platforms.csv guardado correctamente")
```

```python
import pandas as pd

country_to_continent = {
    'USA':       'America',
    'Canada':    'America',
    'UK':        'Europe',
    'Germany':   'Europe',
    'France':    'Europe',
    'Poland':    'Europe',
    'Sweden':    'Europe',
    'Russia':    'Europe',
    'Japan':     'Asia',
    'Australia': 'Oceania'
}

# Actualizar developers.csv
df_devs = pd.read_csv('developers.csv')
df_devs['continent'] = df_devs['country'].map(country_to_continent)
df_devs.to_csv('developers.csv', index=False)
print("developers.csv actualizado")
print(df_devs[['developer_id', 'name', 'country', 'continent']].head(5))

# Actualizar publishers.csv
df_pubs = pd.read_csv('publishers.csv')
df_pubs['continent'] = df_pubs['country'].map(country_to_continent)
df_pubs.to_csv('publishers.csv', index=False)
print("\npublishers.csv actualizado")
print(df_pubs[['publisher_id', 'name', 'country', 'continent']].head(5))
```

```python
import pandas as pd
import random

random.seed(42)

df = pd.read_csv('games2cleaned_10k_new.csv')

rows = []

for _, row in df.iterrows():
    app_id = int(row['appid'])
    
    # Todos tienen Steam
    platforms = [1]
    
    # Epic Games - 40% de probabilidad
    if random.random() < 0.40:
        platforms.append(2)
    
    # GOG - 20% de probabilidad
    if random.random() < 0.20:
        platforms.append(3)
    
    # itch.io - 15% de probabilidad
    if random.random() < 0.15:
        platforms.append(4)
    
    # Amazon Luna - 5% de probabilidad
    if random.random() < 0.05:
        platforms.append(5)

    for platform_id in platforms:
        rows.append({
            'app_id':      app_id,
            'platform_id': platform_id
        })

df_gp = pd.DataFrame(rows)
print(df_gp.shape)
print(df_gp['platform_id'].value_counts())

df_gp.to_csv('game_platforms.csv', index=False)
print("game_platforms.csv guardado correctamente")
```
